```python
import pandas as pd
import os
import glob
```

```python
sensor_name = os.listdir('data/Smart_Citizen_Data')
```

```python
os.getcwd()
```

```python
sensor_dict = {}

for sensor in sensor_name:
    if 'error' in sensor:
        continue
    path = 'data/Smart_Citizen_Data/{}/'.format(sensor)
    dates = glob.glob(os.path.join(path, '*.csv'))
    date_dict = []
    for date in dates:
        date_dict.append(pd.read_csv(date,skiprows=[1,2,3]))
    sensor_dict[sensor] = pd.concat(date_dict, ignore_index=True)
    print('{} files concatenated'.format(sensor))
```

```python
# create a new directory to save the concatenated files
os.mkdir('data/Smart_Citizen_Data_concat')

# save the concatenated files to the new directory
for index in sensor_dict:
    sensor_dict[index].to_csv('data/Smart_Citizen_Data_concat/{}.csv'.format(index))
```
