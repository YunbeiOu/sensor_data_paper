```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from os import listdir
from os.path import isfile, join
```

# Read sensor files

```python
main_folder_path = 'Sensors'

sensor_folders = [folder for folder in os.listdir(main_folder_path) if os.path.isdir(os.path.join(main_folder_path, folder))]

dataframes_by_sensor = {}

for sensor_id in sensor_folders:
    sensor_folder_path = os.path.join(main_folder_path, sensor_id)

    csv_files = [file for file in os.listdir(sensor_folder_path) if file.endswith('.CSV')]

    if len(csv_files) > 1:
        combined_df = pd.DataFrame()

        for csv_file in csv_files:
            csv_file_path = os.path.join(sensor_folder_path, csv_file)
            df = pd.read_csv(csv_file_path,skiprows=[1,2,3])
            combined_df = pd.concat([combined_df, df], ignore_index=True)

        dataframes_by_sensor[sensor_id] = combined_df

    elif len(csv_files) == 1:
        csv_file_path = os.path.join(sensor_folder_path, csv_files[0])
        single_df = pd.read_csv(csv_file_path,skiprows=[1,2,3])

        dataframes_by_sensor[sensor_id] = single_df

    else:
        print(f"No CSV files found for sensor {sensor_id}")

for sensor_id, df in dataframes_by_sensor.items():
    print(f"Data for sensor {sensor_id}:\n{df}")
```

```python
sensor_ids = list(dataframes_by_sensor.keys())
```

## Convert datetime

```python
import pandas as pd

for sensor_id, df in dataframes_by_sensor.items():
    if 'TIME' in df.columns:
        try:
            df['TIME'] = pd.to_datetime(df['TIME'])  
            df['TIME'] = df['TIME'].dt.tz_convert(None)
        except ValueError as e
            print(f"Error processing sensor {sensor_id}: {e}")
```

## Slice the data

```python
import pandas as pd

start_timestamp1 = '2023-12-19 19:00:00'
end_timestamp1 = '2023-12-20 7:00:00'

start_timestamp2 = '2023-12-20 19:00:00'
end_timestamp2 = '2023-12-21 7:00:00'

sliced_dataframes_by_sensor = {}

for sensor_id, df in dataframes_by_sensor.items():
    if 'TIME' in df.columns:
 
        sliced_df = df[((df['TIME'] >= start_timestamp1) & (df['TIME'] <= end_timestamp1))|((df['TIME'] >= start_timestamp2) & (df['TIME'] <= end_timestamp2))]

        sliced_dataframes_by_sensor[sensor_id] = sliced_df
```

# Plot the Temperature and humidity of SCK and HOBO

```python
for sensor_id, df in sliced_dataframes_by_sensor.items():
    fig, ax1 = plt.subplots(figsize=(8, 3)) 
    ax1.plot(df_HOBO['Time'], df_HOBO['Temp'], label='HOBO Temperature', linestyle='-', color='tab:grey')
    ax1.set_xlabel('Time') 
    ax1.set_ylabel('Temperature (°C)')
    ax1.tick_params(axis='y')
    ax1.set_title(f'Recorded Temperature Comparison Between {sensor_id} and HOBO',fontsize=15)
    ax1.grid(True)

    ax1.plot(df['TIME'], df['TEMP'], label=f'{sensor_id} Temperature', linestyle='-', color='tab:blue')


    lines, labels = ax1.get_legend_handles_labels()
    ax1.legend(lines, labels, loc='upper left')

    plt.tight_layout()
    plt.savefig(f'{sensor_id}_HOBO_temperature.png')
    plt.show()
```

```python
for sensor_id, df in sliced_dataframes_by_sensor.items():
    fig, ax1 = plt.subplots(figsize=(8, 3)) 
    ax1.plot(df['TIME'], df['HUM'], label=f'{sensor_id} Relative Humidity', linestyle='-', color='tab:green')

    ax1.plot(df_HOBO['Time'], df_HOBO['Humidity'], label='HOBO Relative Humidity', linestyle='-', color='tab:grey')
    ax1.set_xlabel('Time') 
    ax1.set_ylabel('Relative Humidity (%)')
    ax1.tick_params(axis='y')
    ax1.set_title(f'Recorded Relative Humidity Comparison Between {sensor_id} and HOBO',fontsize=15)
    ax1.grid(True)

    

    lines, labels = ax1.get_legend_handles_labels()
    ax1.legend(lines, labels, loc='upper left')

    plt.tight_layout()
    plt.savefig(f'{sensor_id}_HOBO_Relative Humidity.png')
    plt.show()
```

# Plot the Temperature and humidity differences

```python
df_HOBO_resampled = df_HOBO.set_index('Time')
df_HOBO_resampled = df_HOBO_resampled.resample('5T').mean()
```

```python

resampled_dataframes_by_sensor = {}
for sensor_id, df_sensor in sliced_dataframes_by_sensor.items():
    # Set 'TIME' as the index for resampling
    df_sensor_resampled = df_sensor.set_index('TIME')

    # Resample data to a common frequency
    df_sensor_resampled = df_sensor_resampled.resample('5T').mean()
    
    absolute_difference = df_sensor_resampled['TEMP'] - df_HOBO_resampled['Temp']
    absolute_difference = absolute_difference.abs()
    df_sensor_resampled['abs_diff_temp']=absolute_difference
    
    # Plot the absolute difference
    plt.figure(figsize=(10, 5))
    plt.plot(df_sensor_resampled.index, df_sensor_resampled['abs_diff_temp'], label='Absolute Difference')
    plt.xlabel('Time')
    plt.ylabel('Absolute Temperature Difference (°C)')
    plt.title(f'Absolute Temperature Difference between {sensor_id} and HOBO')
    # plt.ylim(0, 1)
    plt.legend()
    plt.savefig(f'{sensor_id}_temperature_difference.png')

    plt.show()
    
    absolute_difference = df_sensor_resampled['HUM'] - df_HOBO_resampled['Humidity']
    absolute_difference = absolute_difference.abs()
    df_sensor_resampled['abs_diff_hum']=absolute_difference
    resampled_dataframes_by_sensor[sensor_id]=df_sensor_resampled
    
    # Plot the absolute difference
    plt.figure(figsize=(10, 5))
    plt.plot(df_sensor_resampled.index, df_sensor_resampled['abs_diff_hum'], label='Absolute Difference', color='tab:green')
    plt.xlabel('Time')
    plt.ylabel('Absolute Relative Humidity Difference (°C)')
    plt.title(f'Absolute Relative Humidity Difference between {sensor_id} and HOBO')
    plt.legend()
    plt.savefig(f'{sensor_id}_humidity_difference.png')
    plt.show()
    
```
